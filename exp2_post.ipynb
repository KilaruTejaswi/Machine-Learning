```python
# Experiment No: 2
# Preprocessing on Titanic Dataset

import pandas as pd
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler, PowerTransformer

# Load Titanic dataset
df = sns.load_dataset('titanic')

# Display first five records
print(df.head())

# Dataset Information
print("\nDataset Information:")
print(df.info())

# Shape of Dataset
print("\nShape of Dataset:")
print(df.shape)

# Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# -----------------------------
# Data Preprocessing
# -----------------------------

# Fill missing values
df['age'] = df['age'].fillna(df['age'].mean())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])

# Drop deck column (688 missing values)
df.drop('deck', axis=1, inplace=True)

# Missing Values After Imputation
print("\nMissing Values After Imputation:")
print(df[['age','fare','embark_town','embarked']].isnull().sum())

# -----------------------------
# Normalization
# -----------------------------
data = df[['age','fare']]

minmax = MinMaxScaler()
normalized = minmax.fit_transform(data)

print("\nNormalized Data:")
print(normalized[:5])

# -----------------------------
# Standardization
# -----------------------------
standard = StandardScaler()
standardized = standard.fit_transform(data)

print("\nStandardized Data:")
print(standardized[:5])

# -----------------------------
# Power Transformation
# -----------------------------
power = PowerTransformer()
power_data = power.fit_transform(data)

print("\nPower Transformed Data:")
print(power_data[:5])
```
