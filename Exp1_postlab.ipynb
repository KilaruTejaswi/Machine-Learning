```python
# Load dataset directly from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Convert column names to lowercase
df.columns = df.columns.str.lower()

print("First Five Rows:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nDataset Info:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nDescriptive Statistics:")
print(df.describe())
```

```python
df1 = df.copy()

df1["sex"] = df1["sex"].map({"male":0,"female":1})
df1["embarked"] = df1["embarked"].map({"S":0,"C":1,"Q":2})

plt.figure(figsize=(8,6))
sns.heatmap(df1.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap(24EU01088)")
plt.show()
```

```python
fig = plt.figure(figsize=(16,12))

# Survival Count
plt.subplot(3,3,1)
sns.countplot(x="survived", data=df)
plt.title("Survival Count(24EU01088)")

# Gender Distribution
plt.subplot(3,3,2)
sns.countplot(x="sex", data=df)
plt.title("Gender Distribution(24EU01088)")

# Survival by Gender
plt.subplot(3,3,3)
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender(24EU01088)")

# Passenger Class
plt.subplot(3,3,4)
sns.countplot(x="pclass", data=df)
plt.title("Passenger Class Distribution(24EU01088)")

# Age Distribution
plt.subplot(3,3,5)
sns.histplot(df["age"], bins=25, kde=True)
plt.title("Age Distribution(24EU01088)")

# Fare Distribution
plt.subplot(3,3,6)
sns.histplot(df["fare"], bins=25, kde=True)
plt.title("Fare Distribution(24EU01088)")

# Age Boxplot
plt.subplot(3,3,7)
sns.boxplot(x=df["age"])
plt.title("Age Boxplot(24EU01088)")

plt.tight_layout()
plt.show()

# Pairplot
sns.pairplot(df[["age","fare","pclass","survived"]], hue="survived")
plt.show()
```
